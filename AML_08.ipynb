Implementation of One_R

```python
import pandas as pd
rainfall = ["high","high","high","low","low","low"]
soilType = ["loamy","sandy","clay","loamy","sandy","sandy"]
yeild= ["good","good","good","poor","poor","good"]
```

```python

def oneR(df):
    col = list(df.columns)
    target = col[-1]
    length = len(df[target])
    resper = None
    res = ""
    for i in col[:-1]:
        currError = 0
        currCol = set(df[i])
        for j in currCol:
            currGood = df[df[i] == j][target].value_counts()
            currError += currGood.sum()-currGood.max()
        error_rate = currError / length
        if resper is None:
            resper = error_rate
            res = i
        elif error_rate < resper:
            resper = error_rate
            res = i
    print(
        f"The feature {res} is selected "
        f"with error rate of {resper * 100:.4f}%"
    )
data ={
    "rainfall": ["high", "high", "low", "low", "medium", "medium"],
    "soilType": ["clay", "sand", "clay", "sand", "clay", "sand"],
    "yield": ["good", "good", "bad", "bad", "good", "bad"]
}
df = pd.DataFrame(data)
print(df)
oneR(df)
```
