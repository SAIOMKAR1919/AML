Implementation of CFS using pearson corelation

```python
import numpy as np
```

```python
x1 = np.array([1,2,3,7,8,9])
x2 = np.array([2,3,4,6,7,8])
x3 = np.array([5,4,3,3,2,1])
y = np.array([0,0,0,1,1,1])
data = {'x1':x1,'x2':x2,'x3':x3,'y':y}
```

```python
def r(x, y):
    return np.corrcoef(x, y)[0, 1]
```

```python
def CFS_merit(k, x1, x2, y):
    rcf = (abs(r(x1, y)) + abs(r(x2, y))) / 2
    rff = abs(r(x1, x2))

    denominator = k + k * (k - 1) * rff

    if denominator == 0:
        return 0

    return (k * rcf) / np.sqrt(denominator)
```

```python
def best_merit(data):
    cols = [key for key in data.keys() if key != 'y']
    k = 2
    res = {}

    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):
            f1, f2 = cols[i], cols[j]

            score = CFS_merit(
                k, data[f1], data[f2], data['y']
            )

            res[(f1, f2)] = score

    best_pair = max(res, key=res.get)
    return best_pair, res[best_pair], res

best_pair, best_score, results = best_merit(data)

for pair, score in results.items():
    print(f"{pair}: {score:.4f}")

print(f"\nBest feature pair: {best_pair}")
print(f"Best CFS merit score: {best_score:.4f}")
```
