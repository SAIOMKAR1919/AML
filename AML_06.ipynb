Implementation of chi-square.


```python
import numpy as np
o = np.array([
    [10,5],
    [5,10]
])
row_total = np.sum(o,axis=1)
col_total = np.sum(o,axis=0)
grand_total = np.sum(o)
E = np.outer(row_total,col_total)/grand_total
chi_square = np.sum((o-E)**2)/E
critical_value = 3.841
print("Observed: ",o)
print("Expected: ",E)
print("Chi-Square: ",chi_square)
print("Critical value: ",critical_value)
if chi_square > critical_value:
    print("Reject H.")
    print("Sport and buyer are related")
else:
  print("Fail to reject H.")
  print("sports and buyer aren't related")
```
