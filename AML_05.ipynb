Implementation of Lasson and Ridge regression


```python
#Ridge Regression
import numpy as np
import matplotlib.pyplot as plt
def ridge_regression(x,y,learning_rate,iteration,lam):
  n= len(y)
  w=0
  b=0
  cost_history = []
  for i in range(iteration):
    y_pred = w*x+b
    cost = (1/n)*np.sum((y_pred-y)**2)+(lam/2)*(w**2)
    cost_history.append(cost)
    dw = (2/n)*np.sum((y_pred-y)*x)+lam*w
    db = (2/n)*np.sum(y_pred-y)
    w = w-learning_rate*dw
    b = b-learning_rate*db
  return w, b, cost_history
x= np.array([1,2,3,4],dtype = float)
y = np.array([2,3.5,4,5],dtype = float)
learning_rate = 0.05
lam = 0.2
iteration = 1000
w,b,cost_history= ridge_regression(
    x,y,learning_rate,iteration,lam
)
print("Final weight: ",w)
print("Final bias: ",b)
plt.figure()

plt.plot(
    range(len(cost_history)),
    cost_history
)

plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost History vs Iterations")
plt.grid(True)

plt.show()
plt.scatter(x,y,label="Origiaal data")
plt.plot(x,w*x+b,color="red",label="Ridge regression line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Ridge Regression")
plt.legend()
plt.grid(True)
plt.show()
```
