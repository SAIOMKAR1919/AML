**BOX PLOT**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=sns.load_dataset("iris")
print(df.head())

sns.boxplot(y=df["sepal_length"])
plt.title("Box Plot")
plt.ylabel("Length of sepal")
plt.show()
```

**SCATTER PLOT**

```python
import pandas as pd
import matplotlib.pyplot as plt

info={
    "x":[1,5,3,8,5,1,12,6.2,9],
    "y":[30,35,40,45,50,55,60,65,70]
}
df=pd.DataFrame(info)
print(df)

plt.scatter(df['x'],df['y'])
plt.xlabel("Study Hour")
plt.ylabel("Marks")
plt.title("Study V/S Marks")
plt.show()
```

**Histogram**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=sns.load_dataset("titanic")
print(df.head())
print(df["age"].isnull().sum())
age_info=df["age"].dropna()
plt.hist(age_info,bins=10,color='violet')
plt.xlabel("Age")
plt.ylabel("Number of passanger")
plt.title("Titanic passanger age")
plt.show()
```

**Line Chart**

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Date":[
        "2024-01-01",
        "2024-02-01",
        "2024-03-01",
        "2024-04-01",
        "2024-05-01",
        "2024-06-01",
        "2024-07-01",
        "2024-08-01"
    ],
    "Price":[185, 192, 188, 201, 210, 218, 225, 221]
}
df=pd.DataFrame(data)
df['Date']=pd.to_datetime(df['Date'])
plt.plot(df['Date'],df['Price'],marker='o')
plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Stock Price")
plt.show()
```

**Pie Chart**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=sns.load_dataset("titanic")
print(df.head())
class_count=df["pclass"].value_counts().sort_index()
print(class_count)
labels=['first class','second class','Third class']
plt.pie(
    class_count,
    labels=labels,
    startangle=90
)
plt.title("Titanic passanger class")
plt.show()
```

**Bar Plot**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df =sns.load_dataset("penguins")
avg=df.groupby("species")["body_mass_g"].mean()
plt.bar(avg.index,avg.values)
plt.xlabel("Species")
plt.ylabel("Body Mass")
plt.title("Body Mass V/S Penguins")
plt.show()
```
